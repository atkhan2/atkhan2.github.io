```python
import pandas as pd
import altair as alt

alt.data_transformers.disable_max_rows()

url = "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/bfro_reports_fall2022.csv"
df = pd.read_csv(url)

df.head()
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["month"] = df["date"].dt.month
```

```python
plot1 = alt.Chart(df).mark_bar().encode(
    x=alt.X("month:O", title="Month"),
    y=alt.Y("count():Q", title="Number of Sightings"),
    color="month:O"
).properties(
    title="Number of Bigfoot sightings per month"
)
plot1
```

```python
selection = alt.selection_point(fields=["state"], bind="legend")

plot2 = alt.Chart(df).mark_circle(size=60).encode(
    x="longitude:Q",
    y="latitude:Q",
    color=alt.condition(selection, "state:N", alt.value("lightgray"))
).add_params(
    selection
).properties(
    title="Areas for Bigfoot sightings"
)

plot2
```

```python
plot1.save("plot1.html")
plot2.save("plot2.html")
```
